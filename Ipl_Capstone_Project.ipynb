#### Now , Here come some basic questiions

1. Which team won the most matches 

```python
match_wins = df['match_winner'].value_counts()
match_wins
```

```python
match_wins = df['match_winner'].value_counts()
sns.barplot(y = match_wins.index, x = match_wins.values, palette = 'viridis')
plt.title('Most match win by team')
```

2. Toss Decision Treand

```python
sns.countplot(x = df['toss_decision'], palette = 'rainbow')
plt.title('Toss Decision Treand')
```

3. Toss winner & match winner

```python
count = df[df['match_winner'] == df['toss_winner']]['match_id'].count()
persentage = (count * 100)/ df.shape[0]
persentage.round(2)
```

4. How do team win ? (Runs vs wickets)

```python
sns.countplot(x = df['won_by'])
plt.title('Won_by')
```
